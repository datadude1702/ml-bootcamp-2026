```python
name = input ( "What's your name?")

if name == "Harry":
  print ("Gryffindor")
elif name == "Hermione":
  print ("Gryffindor")
elif name == "Ron":
  print ("Gryffindor")
elif name == "Draco":
  print ("Slytherin")
else:
  print ("Who")
```
